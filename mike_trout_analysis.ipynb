```python
df_2015 = pd.read_csv("トラウト2015レギュラーシーズン　savant_data.csv")
df_2019 = pd.read_csv("トラウト2019レギュラーシーズン　savant_data.csv")
df_2022 = pd.read_csv("トラウト2022レギュラーシーズン　savant_data.csv")

print("2015:", df_2015.shape)
print("2019:", df_2019.shape)
print("2022:", df_2022.shape)
```

```python
df = pd.concat(
    [df_2015, df_2019, df_2022],
    ignore_index=True
)

print(df.shape)
```

```python
print(df.columns.tolist())
```

```python
import sqlite3

# SQLiteデータベースを作成
conn = sqlite3.connect(":memory:")

# DataFrameをSQLのテーブルにする
df.to_sql("trout", conn, index=False, if_exists="replace")

print("SQLテーブル作成完了")
```

```python
query = """
SELECT
    pa AS "打席数（PA）",
    ba AS "打率（AVG）",
    obp AS "出塁率（OBP）",
    slg AS "長打率（SLG）",
    (obp + slg) AS "OPS",
    hrs AS "本塁打（HR）",
    hits AS "安打（H）",
    bb AS "四球（BB）",
    so AS "三振（SO）"
FROM trout
WHERE pa = 600
"""

result = pd.read_sql_query(query, conn)

# 小数点以下3桁に統一
for col in ["打率（AVG）", "出塁率（OBP）", "長打率（SLG）", "OPS"]:
    result[col] = result[col].map(lambda x: f"{x:.3f}")

# 表を中央揃え＋枠線＋行番号非表示
display(
    result.style
    .hide(axis="index")
    .set_properties(**{
        "text-align": "center",
        "border": "1px solid black",
        "width": "110px"
    })
    .set_table_styles([
        {
            "selector": "th",
            "props": [
                ("text-align", "center"),
                ("border", "1px solid black"),
                ("width", "110px")
            ]
        },
        {
            "selector": "td",
            "props": [
                ("text-align", "center"),
                ("border", "1px solid black"),
                ("width", "110px")
            ]
        },
        {
            "selector": "table",
            "props": [
                ("border-collapse", "collapse"),
                ("table-layout", "fixed"),
                ("width", "100%")
            ]
        }
    ])
)
```

```python
query = """
SELECT
    hrs,
    barrels_per_bbe_percent,
    hardhit_percent,
    launch_angle
FROM trout
"""

result_3years = pd.read_sql_query(query, conn)

print(result_3years)
```

```python
years = [2015, 2019, 2022]

hardhit_data = [48.6, 44.4, 51.0]

meatball_data = [7.3, 5.3, 6.7]

comparison = pd.DataFrame({
    "年": [f"{year}年" for year in years],
    "本塁打（HR）": result_3years["hrs"].values,
    "バレル率(Barrel%)": result_3years["barrels_per_bbe_percent"].values,
    "ハードヒット率(Hard-Hit%)": hardhit_data,
    "打球角度(Launch Angle)": result_3years["launch_angle"].values,
    "ミートボール率(Meatball%)": meatball_data
})


# 小数点以下1桁に統一
for col in ["バレル率(Barrel%)", "ハードヒット率(Hard-Hit%)", "打球角度(Launch Angle)", "ミートボール率(Meatball%)"]:
    comparison[col] = comparison[col].map(lambda x: f"{x:.1f}")


# 表を中央揃え＋枠線＋行番号非表示
display(
    comparison.style
    .hide(axis="index")
    .set_properties(**{
        "text-align": "center",
        "border": "1px solid black",
        "width": "110px"
    })
    .set_table_styles([
        {
            "selector": "th",
            "props": [
                ("text-align", "center"),
                ("border", "1px solid black"),
                ("width", "110px")
            ]
        },
        {
            "selector": "td",
            "props": [
                ("text-align", "center"),
                ("border", "1px solid black"),
                ("width", "110px")
            ]
        },
        {
            "selector": "table",
            "props": [
                ("border-collapse", "collapse"),
                ("table-layout", "fixed"),
                ("width", "100%")
            ]
        }
    ])
)
```
